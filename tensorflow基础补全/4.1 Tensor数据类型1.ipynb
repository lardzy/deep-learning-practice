*标量*，维度为0，shape为[]

```python
import tensorflow as tf


a = tf.constant(1.6)
print(f'标量：{a}')
```

*向量*，维度为1，shape为[n]

```python
a = tf.constant([1.7, 2, 3.])
print('向量：', a)
print('向量返回Numpy.array类型数据：', a.numpy())
```

*矩阵*，维度为2， shape为[n, m]

```python
a = tf.constant([[1.3, 1.4, 4], [8.9, 91.1, 12], [12, 13.1, 18.9]])
print('矩阵：', a)
```

*三维的张量*

```python
a = tf.constant([[[1.2, 1.3], [1.4, 1.5]],
                 [[1.6, 1.7], [1.8, 1.9]],
                 [[1.9, 2.], [2.1, 2.2]]])
print('维度为3的张量：', a)
```

查看当前使用的设备类型

```python
with tf.device("cpu"):
    a = tf.constant([1])
    print(a.device)

with tf.device("gpu"):
    b = tf.constant([1])
    print(b.device)
```
