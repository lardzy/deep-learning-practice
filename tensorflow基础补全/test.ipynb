```python
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Heiti TC']   # 设置中文字体
plt.rcParams['axes.unicode_minus'] = False           # 解决负号显示问题

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
plt.plot(x, y, color='red', linestyle='--', marker='o', label='线性增长')
plt.title('线性增长图表')
plt.xlabel('x轴')
plt.ylabel('y轴')
plt.legend()
plt.show()
```
