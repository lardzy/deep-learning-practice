## 2.数据结构

### Series（一维）

```python
import pandas as pd

s = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
s
```

### DataFrame（二维）

```python
data = {
    "Name": ["Tom", "Jerry", "Spike", "张三", "李四", "王五"],
    "Age": [20, 22, 24, 34, 18, 22],
    "Score": [88, 92, 95, 88, 78, 98]
}
df = pd.DataFrame(data)
# df.to_excel("test.xlsx", index=False)
df
```

## 5. 数据选择与过滤

### 按列

```python
# 按列
print('取一列（Series）\n', df["Name"])  # 取一列（Series）
df[["Name", "Score"]]  # 取多列（DataFrame）
```

### 按行

```python
# 按行
df.loc[0]  # 按索引标签
```

```python
df.iloc[0]  # 按行号
```

### 条件筛选

```python
df[df['Age'] > 21]
```

```python
df[(df["Age"] > 21) & (df["Score"] > 90)]
```

## 6. 数据清洗

### 缺失值

```python
df.isna().sum()  # 查看缺失值
```

```python
df.dropna()  # 删除含有缺失值的行
```

```python
df.fillna(0)  # 用 0 填充缺失值
```

### 重复值

```python
df.duplicated()  # 查找重复行
```

```python
df.drop_duplicates()  # 删除重复行
```

### 类型转换

```python
df["Age"] = df["Age"].astype(float)
```

## 7. 增删改列

```python
df["Gender"] = ["M", "M", "M", "M", "M", "F"]  # 添加列
```

```python
new_df = df.drop(["Gender", "Score"], axis=1)  # 删除列
new_df
```

```python
new_df = df.rename(columns={"Name": "姓名"})
new_df
```

## 8. 排序与索引

```python
df.sort_values("Age", ascending=False)  # 按 Age 排序
```

```python
df.set_index("Name")  # 设置索引
```

```python
df.reset_index()  # 重置索引
```

## 9. 分组与聚合

```python
df.groupby("Gender")["Score"].mean()   # 按性别求平均分
```

```python
df.groupby("Gender").agg({"Age":"mean","Score":"max"})
```

## 11. 常用统计函数

```python
df["Score"].mean()   # 平均值
```

```python
df["Score"].sum()    # 总和
```

```python
df["Score"].max()    # 最大值
```

```python
df["Score"].min()    # 最小值
```

```python
df["Score"].value_counts()   # 计数统计
```
