1.列表（list）
相当于 Java 的 ArrayList，有序、可变、允许重复元素。

```python
nums = [1, 2, 3, 4]
nums.append(5)
nums.remove(2)  # 删（按值）
print(nums[0])
nums[1] = 10  # 改（按位置）
print(nums)
```

## 1.列表
创建一个列表 nums = [10, 20, 30, 40, 50]：

输出第一个和最后一个元素

在末尾加上一个元素 60

删除元素 30

把第二个元素改成 25

```python
nums = [10, 20, 30, 40, 50]
print('1 -> 第一个元素：', nums[0], '\n最后一个元素：', nums[-1])
nums.append(60)
print('2 -> 在末尾加上一个元素60:', nums)
nums.remove(30)  # 删（按值）
print('3 -> 移除元素30：', nums)
nums[1] = 25
print('4 -> 将第二个元素改为25:', nums)
```

给定 nums = [1, 2, 2, 3, 4, 4, 5]，去掉重复元素，输出新的列表。

```python
nums = [1, 2, 2, 3, 4, 4, 5]
result = []
for num in nums:
    if num not in result:
        result.append(num)

print('result:', result)
```

定义一个元组 point = (3, 4)，输出横坐标和纵坐标。

创建两个集合：
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
求并集
求交集
求差集

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print('并集：', a | b)
print('交集：', a & b)
print('a - b 差集：', a - b)
print('b - a 差集：', b - a)
```

4. 字典（dict）

定义一个字典：

student = {"name": "Tom", "age": 18, "gender": "M"}


输出 name 的值

把 age 改成 20

添加一个新的键值对 "score": 90

遍历字典，打印每一对 key 和 value

```python
students = {"name": "Tom", "age": 18, "gender": "M"}
print(students.get("name"))
```

```python
students["age"] = 20
print(students)
```

```python
students["score"] = 90
for k,v in students.items():
    print(f'key:{k},value:{v}')
```

```python
del students["gender"]
print(students)
```

# 三、函数与控制结构

写一个函数 is_even(n)，判断一个数是否是偶数。

```python
def is_even(n: int) -> bool:
    if n % 2 == 0:
        return True
    else:
        return False


print(is_even(2))
```
