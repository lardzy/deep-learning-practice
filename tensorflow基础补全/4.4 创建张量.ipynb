# 创建张量，使用tf.convert_to_tensor()

## 1.从列表创建张量

```python
import tensorflow as tf
import numpy as np

a = tf.convert_to_tensor([1, 2.1, 3.])
a
```

## 2.从数组创建张量  
**需要注意的是 Numpy 浮点数数组默认使用 64 位精度**

```python
a = tf.convert_to_tensor(np.array([[1.2, 2.3], [1., 4.3]]))
print('从数组创建张量:', a, '\n数据精度:', a.dtype)
```

## 3.创建全0或者全1的张量

```python
a = tf.zeros([2, 3])
print('\n全0张量:', a)
a = tf.ones([5])
print('\n全1张量', a)
```

## 4.通过tf.zeros_like、tf.ones_like创建与某个张量shape一致的全零或者全1张量

```python
a = tf.constant([[1.2, 1.4, 1.], [5., 4.1, 2.9]])
print('a.shape:', a.shape)
a = tf.ones_like(a)
print('\nshape一致全1张量:', a)
```

## 5.创建自定义数值的张量，使用tf.fill()

```python
a = tf.fill([3, 4], 6)
print('\n自定义数值的张量:', a)
```

## 6.创建已知分布的张量  
正态分布(Normal Distribution，或 Gaussian Distribution)和均匀分布(Uniform Distribution)是最常见的分布之一，  
创建采样自这 2 种分布的张量非常有用，比如在卷积神经网络中，*卷积核张量 𝑾 初始化为正态分布有利于网络的训练*；  
在对抗生成网络中，隐藏 变量𝒛一般采样自均匀分布。


```python
# 标准正态分布张量
a = tf.random.normal([3, 2])
print('标准正态分布:', a)
```

```python
# 创建均值为1，标准差为2的正态分布
a = tf.random.normal([2, 2], mean=1, stddev=2)
print('创建均值为1，标准差为2的正态分布:', a)
```

```python
# 创建采样自 [ minval , maxval ) 区间的均匀分布的张量
# 通过 tf.random.uniform(shape, minval=0, maxval=None, dtype=tf.float32)
a = tf.random.uniform([16, 2], minval=0, maxval=2)
print('区间均匀分布张量:', a)
```

```python
# 指定区间中均匀采样整形类型张量
a = tf.random.uniform([3, 2],maxval=100, dtype=tf.int32)
print('整形均匀分布张量:', a)
```

```python
# 创建序列
# 通过 tf.range(start, limit, delta=1)可以创建 [ start , limit ) ，步长为 delta 的序列，不包含 limit 本身
a = tf.range(1, 100, 2)
print('[1, 100)步长为2的序列:', a)
```
