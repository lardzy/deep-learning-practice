```python
import tensorflow as tf


a = tf.random.normal([4, 32, 32, 3])
# 取一张图片的数据
# print('a[0]', a[0])
# 取一张图片第一行的数据
print('a[0][0]', a[0][0].shape)
# 取第 2 张图片，第 10 行，第 3 列的数据
print('a[1][9][2]', a[1][9][2].shape)
# 当张量的维度数较高时，使用[𝑖][𝑗]. . .[𝑘]的方式书写不方便，可以采用[𝑖,𝑗, … , 𝑘]的方式索引，它们是等价的
# 取第 2 张图片，第 10 行，第 3 列的数据
print('取第 2 张图片，第 10 行，第 3 列的数据:', a[1, 9, 2].shape)
```

# 2.切片

通过start: end: step切片方式可以方便地提取一段数据，其中 start 为开始读取位置的索引，end 为结束读取位置的索引* (不包含 end 位) *，step 为采样步长。

```python
# 以 shape 为[4,32,32,3]的图片张量为例，我们解释如果切片获得不同位置的数据。例如
# 读取第 2,3 张图片
x = tf.random.normal([4, 32, 32, 3])
x[1:3].shape
```

start: end: step切片方式有很多简写方式，其中 start、end、step 3 个参数可以根据需要
选择性地省略，*全部省略时即为::*，表示从最开始读取到最末尾，步长为 1，即不跳过任何
元素。  
如 x[0,::]表示读取第 1 张图片的所有行，其中::表示在行维度上读取所有行，它等价
于 x[0]的写法

```python
x[0, ::].shape  # 代表第一张图片
```

为了更加简洁，::可以简写为单个冒号:

```python
x[:, 0:28:2, 0:28:2, :].shape  # 示读取所有图片、隔行采样、隔列采样，、读取所有通道数据，相当于在图片的高宽上各缩放至原来的 50%
```

特别地，step 可以为负数，考虑最特殊的一种例子，当step = −1时，start: end: −1表
示从 start 开始，逆序读取至 end 结束(不包含 end)，索引号𝑒𝑛𝑑 ≤ 𝑠𝑡𝑎𝑟𝑡。

```python
# 考虑一个 0~9 的
# 简单序列向量，逆序取到第 1 号元素，不包含第 1 号
x = tf.range(9)
print('逆序采样：', x[::-1])
```

```python
# 逆序间隔采样
print('逆序间隔采样:', x[::-2])
```

```python
# 读取每张图片的所有通道，其中行按着逆序隔行采样，列按着逆序隔行采样
x = tf.random.normal([4, 32, 32, 3])
print('读取每张图片的所有通道，其中行按着逆序隔行采样，列按着逆序隔行采样 -> ', x[:, ::-2, ::-2].shape)
```

当张量的维度数量较多时，不需要采样的维度**一般用单冒号:表示采样所有元素**，此时
有可能出现大量的:出现。继续考虑[4,32,32,3]的图片张量，当需要读取 G 通道上的数据
时，前面所有维度全部提取，此时需要写为  
**注意区分索引和切片！！！**


```python
print('G通道：', x[..., 1].shape)
```

为了避免出现像 [: , : , : ,1]这样过多冒号的情况，*可以使用⋯符号表示取多个维度上所有的数据*，其中维度的数量需根据规则自动推断：  
当切片方式出现⋯符号时，⋯符号左边
的维度将自动对齐到最左边，⋯符号右边的维度将自动对齐到最右边，此时系统再自动推
断⋯符号代表的维度数量，

```python
# 读取第 1~2 张图片的 G/B 通道数据
print('读取第 1~2 张图片的 G/B 通道数据:', x[1:3,:,:,1:3].shape)
# 读取最后 2 张图片
print('读取最后 2 张图片:', x[-2:].shape)
# 读取 R/G 通道数据
print('读取 R/G 通道数据:', x[...,:2].shape)
```

张量的索引与切片方式多种多样，尤其是切片操作，初学者容易犯迷糊。但本质上切
片操作只有start: end: step这一种基本形式，通过这种基本形式有目的地省略掉默认参数，
从而衍生出多种简写方法，这也是很好理解的。它衍生的简写形式熟练后一看就能推测出
省略掉的信息，书写起来也更方便快捷。由于深度学习一般处理的维度数在四维以内，⋯
操作符完全可以用:符号代替，因此理解了这些就会发现张量切片操作并不复杂。
