# 创建指定精度张量
创建张量时，可以指定张量的保存精度

```python
import tensorflow as tf
import numpy as np

a = tf.constant(1234567, dtype=tf.int16)
print('int16:', a)  # 精度溢出
a = tf.constant(1234567, dtype=tf.int32)
print('int32:', a)
```

采用tf.float32\tf.float64保存π

```python
a = tf.constant(np.pi, dtype=tf.float32)
print(a)
a = tf.constant(np.pi, dtype=tf.float64)
print(a)
```

# 读取精度

通过访问张量的D-type成员属性可以判断张量的保存属性

```python
# 通过tf.cast函数可以完成精度的转换
print('before:', a.dtype)
if a.dtype != tf.float32:
    a = tf.cast(a, tf.float32)
print('after:', a.dtype)
```

# 类型转换  
低精度转换为高精度

```python
a = tf.constant(np.pi, dtype=tf.float16)
a = tf.cast(a, dtype=tf.float64)
print('低精度转换为高精度:', a)
```

高精度张量转换为低精度张量（数据溢出）

```python
a = tf.constant(1234567, dtype=tf.int64)
a = tf.cast(a, dtype=tf.int16)
print('高精度转为低精度(溢出):', a)
```

*布尔类型转换为整数类型*

```python
a = tf.constant([True, False, True])
a = tf.cast(a, dtype=tf.int32)
print('布尔类型转换为整数类型:', a)
```

整数类型转换为布尔类型

```python
a = tf.constant([1, 0, 2])
a = tf.cast(a, dtype=tf.bool)
print('整数类型转换为布尔类型:', a)   # Tensorflow中，任何非0数字都视为True
```
